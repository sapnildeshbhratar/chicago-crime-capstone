Original CSV(chicago_crime_dataset.csv) has a formatting problem in the last two records:
- Line 498: the location field starts with a quotation mark but does not close it.
- Line 499: the next crime record is also attached to the same malformed quoted field.
- The two records have been merged into one physical CSV line.

Adding missing closing quotation mark to 498 location fields.

Save a repaired copy in data\processed.

Leave the original CSV untouched.

```python
from pathlib import Path

# Original and repaired file paths
original_file = RAW_DATA_DIR / "chicago_crime_dataset.csv"

PROCESSED_DATA_DIR = PROJECT_DIR / "data" / "processed"
PROCESSED_DATA_DIR.mkdir(parents=True, exist_ok=True)

repaired_file = (
    PROCESSED_DATA_DIR / "chicago_crime_dataset_repaired.csv"
)

# Read original file
with open(original_file, "r", encoding="utf-8-sig") as file:
    lines = file.readlines()

# Case numbers of the two records identified
problem_cases = ["HZ251541", "HZ251540"]

repaired_cases = []

for i, line in enumerate(lines):

    if any(case in line for case in problem_cases):

        # Check whether the location field has an unmatched quote.
        if line.count('"') % 2 != 0:
            lines[i] = line.rstrip("\r\n") + '"\n'

            for case in problem_cases:
                if case in line:
                    repaired_cases.append(case)

# Save repaired copy
with open(repaired_file, "w", encoding="utf-8-sig", newline="") as file:
    file.writelines(lines)

print("Repaired file saved:", repaired_file)
print("Records repaired:", repaired_cases)
```

Verify the repaired CSV

```python
import csv

with open(
    repaired_file,
    "r",
    encoding="utf-8-sig",
    newline=""
) as file:

    reader = csv.reader(file)

    header = next(reader)

    expected_columns = len(header)
    invalid_rows = []

    for row in reader:
        if len(row) != expected_columns:
            invalid_rows.append({
                "line": reader.line_num,
                "fields_found": len(row)
            })

print("Expected columns:", expected_columns)
print("Invalid rows:", invalid_rows)
print("CSV validation:", "PASSED" if not invalid_rows else "FAILED")
```
